Datasets Covered

Before deciding on the dataset to use , we had a look at other open source options relate to spaced repetition and learning i.e

1. Anki review logs : This had around 10k users from Hugging Face.It was approximately 15.7GB.It focused on spaced repetition app data and a self rated recall structure of 1-4.

2. FSRS-Anki - Similar to the Anki review logs above with around 20k users and 20GB in size larger than the previous one

3. Duolingo half-life dataset (chosen) : It exists in the Havard Dataverse with a size of 361 MB .It has 12.8M rows and has language learning practice sessions

4. EdNet (KT1) : This is from an online quiz/tutoring app by Riiid Labs.It has 131 million interactions and the quiz style in it is in form of correct /incorrect answers

5. ASSISTments 2009/2012 : This is a public math tutoring benchmark .The actual size wasnt as available but it is smaller than the others and also contains a quiz style of correct/incorrect

6. Junyi Academy : The dataset is in kaggle with around 16 Millions attempts and the quiz style is correctness based

7. OULAD :This is from Open University's learning-analytics records of real students and its based on pass or fail outcomes.

We also came across several Kaggle "study habits vs exam score" datasets exist but they were rejected because they are synthetic (generated by a script or LLM, not real recorded behavior) hence not usable as evidence of real learning patterns.

Output variable per dataset

1. Anki review logs : output variable is a rating, a self rated score (1-4) that the learner gives themselves

2. FSRS-Anki-20k : It contains the same idea as the first one for the rating column

3. Duolingo (chosen) : Output variable is p_recall, the fraction of correct recalls in a session and we already verified it against the actual loaded file.

4. EdNet : The output variable was whether the quiz answer was correct or incorrect and is stored as a 0/1

5. ASSISTments : It has the same idea for correct column as the EdNet and it is also stored as a 0/1

6. Junyi Academy : Again also has the same idea as the EdNet by checking the correctness of the attempt

7. OULAD : The output variable is final_result and it checks whether the student passed,failed or withdrew from the module
